```python
import numpy as np
a = np.array([3.0,4.0])
b = np.array([5.0,6.0])
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
dot_ab=np.dot(a,b)
print(norm_a)
print(norm_b, dot_ab)
```

```python
def cosin_simularity(a:np.ndarray,b:np.ndarray)->float:
    denom = np.linalg.norm(a)*np.linalg.norm(b)
    if denom == 0:
        return 0
    return float(np.dot(a,b)/denom)
```

```python
#Посчитай вручную на бумаге норму вектора [1, 2, 2].
f = np.array([1,2,3])
norm_f= np.linalg.norm(f)
print(norm_f)
```

```python
#Для u = [1, 0], v = [0, 1] посчитай скалярное произведение и косинусную близость.
u = [1,0]
v = [0,1]
print(cosin_simularity(u,v))
print(np.dot(u,v))
#Потому что по формуле косинусной близости у нас 0 * 1 и все дает 0
#Скалярное произведение так же дает 1 * 0 + 0 * 1 = 0
```

```python
#Для u = [1, 0], w = [-1, 0] сделай то же самое.
u = [1,0]
v = [-1,0]
print(cosin_simularity(u,v))
print(np.dot(u,v))
#Потому что по формуле косинусной близости у нас -1 * 1 и все дает -1
#Скалярное произведение так же дает -1 * 1 + 0 * 1 = -1
```
